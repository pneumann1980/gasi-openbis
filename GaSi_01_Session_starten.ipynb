# 1 · Session starten

Hier steht der **einzige Schreibschalter** für alle GaSi-Notebooks.
`SCHREIBEN = False`: Vorschau. `SCHREIBEN = True`: ausgeführte Arbeitsnotebooks schreiben in openBIS.
Der Modus wird lokal in `.env` gespeichert und von jedem Notebook frisch gelesen.
Er bleibt bis zur nächsten Änderung hier bestehen, auch nach einem Kernel-Neustart.
Zum Beenden des Schreibmodus nur die Schalterzelle mit `False` erneut ausführen.

Die Token-Zelle fragt direkt und verdeckt mit getpass nach dem Session-Token.
Leere Eingabe behält den vorhandenen Token. Credentials und Modus werden nicht nach Git übernommen.


```python
import gasi_session

SCHREIBEN = False
print(gasi_session.start_session(SCHREIBEN))
```

```python
import os
from getpass import getpass
import gasi_session

token = getpass("Playground Session Token: ").strip()
try:
    print(gasi_session.save_playground_token(token))
    if token:
        os.environ.pop(gasi_session.TOKEN_ENV, None)
finally:
    del token
print(gasi_session.session_status())
```
