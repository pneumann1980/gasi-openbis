# 01 · Datenmodell und farbigen Forschungsablauf erzeugen

**Aufgabe:** Das aktuelle fachliche Modell und den Sensor-Prüfablauf nach der BAM-Vorlage öffnen.

Mit frischem Kernel im Repository **Run All** ausführen. Der Link
**Farbigen Forschungsablauf öffnen** führt zu den vier BAM-Schritten mit
Sensor, Belastungsprofil, farbigen Boxen und Typentabellen.
Zusätzlich entstehen die einfache Geräteübersicht, technische Details und vier
separat nutzbare SVG-Dateien.

Quellen: `docs/GaSi_Konzeptmodell.md` und `docs/GaSi_Datenmodell.md`.
Geräteunterlagen und Messwerte werden nicht eingebettet.
Falls nötig: `%pip install -r requirements-docs.txt` im aktiven Kernel.
Nach Modelländerungen erneut ausführen; die Erzeugungszelle lädt den Generator neu.
Kein Netzwerk und kein BAM-Zugang erforderlich.

```python
from pathlib import Path
from IPython.display import FileLink, display

model_source = Path("docs/GaSi_Datenmodell.md")
```

```python
# Auch beim erneuten Ausführen nur dieser Zelle aktuellen Generator verwenden.
import importlib
import gasi_model_html

importlib.invalidate_caches()
gasi_model_html = importlib.reload(gasi_model_html)

try:
    html_path = gasi_model_html.generate_simple_html(model_source)
except ImportError:
    print("HTML-Paket fehlt. In einer Notebook-Zelle ausführen:")
    print("%pip install -r requirements-docs.txt")
    print("Danach diese Zelle erneut ausführen.")
else:
    relative_path = html_path.relative_to(Path.cwd().resolve()).as_posix()
    print(f"EINFACHE ANSICHT erzeugt: {relative_path}")
    display(FileLink(relative_path, result_html_prefix="Einfache Grafik öffnen: "))
    concept_path = model_source.with_name("GaSi_Konzeptmodell.html")
    if concept_path.is_file():
        display(FileLink(concept_path.as_posix(), result_html_prefix="Farbigen Forschungsablauf öffnen: "))
```
