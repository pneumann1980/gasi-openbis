# 02 · Aktuelles Setup prüfen

**Aufgabe:** Prüfen, ob die aktuell verwendeten Geräte, Positionen und Kanäle nachvollziehbar beschrieben sind.

Frischer Kernel im Repository, dann **Run All**. Gelesen wird ausschließlich die lokale,
von Git ausgeschlossene Datei `config/gasi_setup.json`. Ein lokaler HTML-Bericht zeigt
die Geräte und die noch offenen Angaben. Es werden keine Geräte in openBIS angelegt.
Eine Umbauhistorie ist nicht Teil dieses Schritts.

Fehlt die Datei: `config/gasi_setup.example.json` nach `config/gasi_setup.json` kopieren.
Felder der Geräte: `manufacturer`, `model`, `serial_number`, `inventory_id`,
`installed_confirmed`, `evidence`, `open_issues`, `calibration_status`.
Für Durchfluss zusätzlich `range_max`, `unit`, `quantity=normal_volume_flow`,
`reference_temperature_C`, `reference_pressure_hPa_abs`, `design_gas`,
`calibration_medium` und `operation_gas`. Eine Bestätigung wird nicht aus Unterlagen abgeleitet.

```python
from pathlib import Path
from IPython.display import FileLink, display
from gasi_setup import inspect_setup, show_setup_summary, write_setup_report

repo_root = Path.cwd()
try:
    setup_report = inspect_setup(repo_root / "config/gasi_setup.json")
except ValueError as exc:
    setup_report = None
    print(f"Setup-Konfiguration prüfen: {exc}")
if setup_report is not None:
    show_setup_summary(setup_report)
    local_html = write_setup_report(setup_report, repo_root / "data/datasheets/_auswertung/GaSi_Aktuelles_Setup.html")
    if local_html is not None:
        display(FileLink(local_html.relative_to(repo_root).as_posix(), result_html_prefix="Aktuelles Setup öffnen: "))
```

Der Bericht bleibt mit den Geräteunterlagen lokal. Die Notebook-Ausgabe zeigt nur
eine Zusammenfassung zur Vollständigkeit, keine Seriennummern oder Beleginhalte.
`installed_confirmed=true` erst setzen, wenn die Zuordnung zum aktuellen Aufbau bestätigt ist;
`calibration_status="assessed"` bezeichnet eine erfolgte Bewertung durch das Prüfmittelmanagement.
Die lokale Prüfung bestätigt keine openBIS-Typen, Pflichtfelder oder Schreibrechte.
