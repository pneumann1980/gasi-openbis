# 00 · Session zentral einrichten

**Aufgabe:** Den openBIS-Session-Token einmal verdeckt eingeben und lokal speichern.

Mit frischem Kernel **Run All** ausführen. Den Token im BAM-Playground unter
**Tools → User Profile** kopieren. Die Eingabe wird nicht angezeigt.
Gespeichert wird ausschließlich in der von Git ausgeschlossenen `.env` im Repository.
Diese lokale Datei enthält die Werte im Klartext; Notebook-Ausgaben enthalten sie nicht.

Notebook 04 und 05 lesen die Datei automatisch. Eine leere Eingabe behält den bisherigen
Token. Weitere Sessionwerte können als `NAME=WERT` in derselben `.env` stehen.
Nichtleere Umgebungsvariablen haben Vorrang vor der Datei. Dieses Notebook verbindet
sich nicht mit openBIS und verändert keine Forschungsdaten.

```python
from getpass import getpass
from gasi_session import save_playground_token, session_status, SessionConfigError

print(session_status())
try:
    session_token = getpass("Playground Session Token: ").strip()
except (EOFError, NotImplementedError, KeyboardInterrupt):
    session_token = ""
    print("Eingabe hier nicht möglich/abgebrochen. Notebook interaktiv ausführen oder lokale .env bearbeiten.")
try:
    print(save_playground_token(session_token))
except SessionConfigError as exc:
    print(str(exc))
finally:
    session_token = ""
    del session_token
```

Danach **Notebook 04** zum Lesen beziehungsweise **Notebook 05** für den
standardmäßig deaktivierten synthetischen Schreibtest öffnen. Nach Ablauf des Tokens
hier einen neuen eingeben und dort die Verbindungszelle erneut ausführen.
Die Sessiondatei wird bei jedem Verbindungsversuch neu gelesen; es gibt keine Token-Kopie
in den anderen Notebooks und keinen automatischen Logout.
